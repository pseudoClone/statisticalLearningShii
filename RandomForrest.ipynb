# Random Forest Algorithm(Use XGBoost or CatBoost or LightGBM for speed, performance tradeoff)

- Ensemble learning methods since it mixes a lot of decisions trees by creating multiple bootstrap datasets.
- At each split in the decision tree, a random feature is selected.

> Let the dataset at hand be $D = (x_1, y_1), (x_2, y_2), \cdots, (x_n, y_n) $ with $N$ samples and $M$ features

### Step 1
**Boostrap Aggregation(Bagging)**
For each of the $B$ trees that we form,

- Sample N examples with replacement from the original dataset dataset to form a dataset $D_b$
- Some examples will repeat and some will be left out(Out of Bag Samples)
### Step 2
**Tree Construction**
For each split in the tree

- Select a random subset of $m$ features out of $M$ features. Or more generally, $m = \sqrt{M}$ or $m = \frac{M}{3}$
- Among m features, pick the best one based on splitting criterion, i.e, Gini Impurity or Information Gain

### Step 3 
**Prediction**
For classification: Each tree votes, and majority class wins.

$\hat{y} = mode({T_1(x), T_2(x), \cdots, T_B(x)})$

$\hat{y} = \frac{1}{B}\sum\limits_{b = 1}^{B} T_b(x)$
