# Single Value Decomposition

- $A = U\sum V^{T}$
- $U$ = $m \times n$ orthogonal matrix whose columns are left singular vectors of A
- $\sum$ = A diagonal $m \times n$ matrix containing the singular values of A in descending order
- $V^T$ = The transpose of n $\times$ n orthogonal matrix where the columns are the right singular vectors of A
- $A$ is $m \times n$

## Steps
#### Step 1: Compute $AA^T$
#### Step 2: Find eigen value of $AA^T$
#### Step 3: Find the right singular vectors (Eigen vectors of $A^TA$)
#### Step 4: Compute left singular vector (Vector $U$)
#### Step 5: Find the SVD equation
