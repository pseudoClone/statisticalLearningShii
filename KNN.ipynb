# K-Nearest Neighbours

K-NN does not learn the model in the ways that we have studied statistical learning methods previously. It memorizes the dataset and only makes predictions when queried. This feature of K-NN is called lazy learning. 

Mostly useful in cases where dataset is small and relatively unfamiliar dataset.

The core idea is to compute distance between data points. If the feature space is continuous and equally scaled, use Euclidian distance. If the feature space is highly sparse, use Manhattan distance.

The main point being, K(number of neighbours) is the only hyperparameter in this learning. For high value of K, the variance is high and bias is low. And its opposite the other way.

To the K nearest neighbours, if uniform weighting is used, $w_i$ = 1. For other weighting schemes, the weights are given as some mathamatically sound formula.

The most common weighting scheme is inverse distance weighting. The closer the instance, the higher the weight.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
y_binary = np.where(y == 0, 1, 0)

X_train, X_test, y_train, y_test = train_test_split(X, y_binary, random_state=69, test_size=0.3)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))
```

```python
def knn_predict(X_train, y_train, X_test, k=3):
    predictions = []
    for test_point in X_test:

        distances = [euclidean_distance(test_point, train_point) for train_point in X_train]
        #print(distances)
        
        k_indices = np.argsort(distances)[:k]
        print(k_indices)
        
        k_nearest_labels = y_train[k_indices]
        
        # Majority Voting, Most common labels
        most_common = np.bincount(k_nearest_labels).argmax()
        predictions.append(most_common)
    
    return np.array(predictions)
```

```python
def accuracy(y_true, y_pred):
    return np.sum(y_true == y_pred) / len(y_true)

k = 3
y_pred = knn_predict(X_train, y_train, X_test, k)

acc = accuracy(y_test, y_pred)
print(f"Accuracy: {acc * 100:.2f}%")
```
