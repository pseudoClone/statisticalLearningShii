# Cost Function

For improved estimation, we want to have values of w, b that we can obtain programatically, so that it fits all $x^{(i)}, y^{(i)}$

Lets get the cost function step by step:
1. $J(w, b) = $ $y - \hat{y}$. This is the error i.e difference between actual value and predicted value. We want this to be as low as possible.
2. Now let's do this for all the training examples.$J(w, b) = $ $\sum\limits_{i = 1}^{m} (y^{(i)} - \hat{y}^{(i)}$
3. This may get larger as the number of training examples increases, so we balance this out with its average. That is:  $J(w, b) = $ $\frac{1}{n}\sum\limits_{i = 1}^{m} (y^{(i)} - \hat{y}^{(i)})$
4. Now the sum may not always be positive. And having negative values in the sum makes life difficult for us. So we make sure the sum is always positive. And we do that by squaring the sum i.e. $J(w, b) = $ $\frac{1}{m}\sum\limits_{i = 1}^{m}(y^{(i)} - \hat{y}^{(i)})^2$
5. An even better cost function is: $J(w, b) = $ $\frac{1}{2m}\sum\limits_{i = 1}^{m}(y^{(i)} - \hat{y}^{(i)})^2$

Now lets visualize the cost function in linear regression with the representation given above.

We have the model: $ f_{a,b}(x) = wx + b $. The parameters are: $w$ and $b$. The cost function is: $J(w, b) = $ $\frac{1}{2m}\sum\limits_{i = 1}^{m}(y^{(i)} - \hat{y}^{(i)})^2$ . Our goal is to minimize $J(w,b)$.

For simplicity, let's set $b = 0$. Then, $f_{w}(x) = wx$. And $J(w) =$ $ \frac{1}{2m} \sum\limits_{i = 1}^{m}(f_{w}(x^{(i)}) - y^{(i)})^2$

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    f_wb = w * x + b
    cost = (f_wb - y) ** 2
    total_cost = (1 / (2 * m)) * np.sum(cost)
    return total_cost

x_train = np.array([1.0, 1.7, 2.0, 2.5, 3.0, 3.2])
y_train = np.array([250, 300, 480,  430,   630, 730])
```

```python
w = np.random.randn()
b = np.random.randn()

# Compute cost with initial w and b
z = compute_cost(x_train, y_train, w, b)

w_values = np.linspace(-10, 10, 100)  # Range of w values
cost_values = [compute_cost(x_train, y_train, w_val, b) for w_val in w_values]
print(cost_values)
```

```python
plt.plot(w_values, cost_values, c='b', label="Cost Function Output")
plt.xlabel('Weight (w)')
plt.ylabel('Cost')
plt.title('Cost Function vs. Weight')
plt.legend()
plt.show()
```
