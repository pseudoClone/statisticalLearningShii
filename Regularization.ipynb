# Regularization

It's just a technique used for preventing overfitting in machine learning models.

The idea is to just add a penalty to the model's objective function/loss function.

In the case of linear regression, the original cost function is the mean squared error function i.e

$\Large{L(w) = \frac{1}{N}\sum\limits_{i = 1}^{N} {(y_i - \hat{y})^2}}$

Now adding L2 Ridge regularization, 

$\Large{L_{ridge} = \frac{1}{N} \sum\limits_{i = 1}^{N}(y_i - \hat{y})^2 + \lambda\sum\limits_{j = 1}^{d}(w_j^2)}$

Now, the gradient of $L_{ridge}$ is:

$\nabla_wL_{ridge}$ = $\frac{2}{N}(y_i - \hat{y}_i) * (-X_{ik}) + 2\lambda  * w_k$

Note that this is differentiated with respect to all the w not just one, so it's essetially $w_k$ or $w_d$

Now this turns to,

$\nabla_w L_{ridge}$ = $\frac{2}{N} X ^ T (Xw - y)$ +  $2\lambda  * w_k$

- For, L1 Norm, just use the Manhattan norm in the regularization term instead of the squares of the weights vector.

- The difference being, this leads to weight components being driven to zeros as the $\lambda$ increases. So, the model sparsity increases. Why is this important? Because this leads to automatic feature selection. Just think, the model just drops the inputs with zeros weights, meaning it has not importance to the output. So, the model becomes lighter and faster.

- The reason why L2 is called smooth shrinkage is because, it smoothly decays the weights unlike L1 which just snaps to zero. This is true because $w^2$ is a polynomial and differentiable for all w
