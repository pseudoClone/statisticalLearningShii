# Backpropagation(For classification neural networks)

The basic idea for this is that, the network will have input, hidden and output layers.

In forward propagation, the network makes predictions using the current set of weights and biases.
Each layer computes the following:

$$\Large{z^{[L]} = W^{[L] a ^{[L-1]} + b^{[L]}}}$$

Where,
>$\Large{z^{[L]}}$ is the weighted sum of the inputs to L

>$\Large{W^{[L]}}$ is the weights matrix for layer L

>$\Large{a^{L - 1}}$ is the output of the previous layer

>$\Large{b^{[L]}}$ is the bias for layer l


After computing $\Large{z^{[L]}}$, the activation function $\Large{g}$ is applied to get the output of that layer:

$$\Large{a^{[L]} = g(z^{[L]})}$$



## Error Calculation
For binary classification, the loss function is binary cross entropy loss or log loss.

$$C_0(y, \hat{y}) = - \left[ y \log(\hat{y}) + (1 - y) \log(1 - \hat{y}) \right]$$

Where,

y is the true label

$\hat{y}$ is the predicted label

---


Now, for a given loss we want to calculate, how much each weight contributed to generating the additive error. That is, we want to compute the gradients of the loss function with respect to each weights. The gradients are then used to update the weights in the direction that minimize the loss.


## Backpropagation

We want to compute the error at the output layer. This is because we want to see, how much small a change in output affects the loss.

$$ \frac{\partial{C_0}}{\partial{a^{[L]}}}  = \hat{y} - y$$

$$ \frac{\partial{C_0}}{z^{[l]}} = \frac{\partial{C_0}}{\partial{a^{[L]}}} \cdot g^{'}(z^{[L]}) $$

We know that:

$\Large{z^{[L+1]} = W^{[l + 1]} a^{[L]} + b^{L+1}}$

By chain rule,

$\Large{\frac{\partial{L}}{\partial{a^{[L]}}} = W^{L + 1} \cdot \frac{\partial{C_0}}{\partial{z^{[L+1]}}}}$
