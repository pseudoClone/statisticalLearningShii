# Linear Regression with Manual Weights and Biases
Here we use manual values of w and b instead of a cost function to create a model to make estimate for the house prices

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1.0, 2.0])
y_train = np.array([300, 500])
```

```python
print(f"X_Train = {x_train}, Shape = {x_train.shape}")
print(f"Y_Train = {y_train}, Shape = {y_train.shape}")
```

```python
m = x_train.shape[0] # Number of training examples
print(m)
```

**Now, we know that the linear regression function is just $f_{w,b}(x) = wx^{(i)} + b$**

**For our cases, since we are not using a cost function just yet, we use w = $100$, b = $100$**

```python
w = 200
b = 100
```

```python
def estimateModel(x, w, b):
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x [i] + b
    return f_wb
    
```

```python
computedF_WB = estimateModel(x_train, w, b)
plt.plot(x_train, computedF_WB, c = 'b', label = "Model Prediction")
plt.scatter(x_train, y_train, marker = 'x', c = 'r', label = "Training Values/Actual Values")
plt.ylabel("House Prices in $1000s")
plt.xlabel("House Area in 1000 sq. ft")
plt.legend()
plt.show()
```
