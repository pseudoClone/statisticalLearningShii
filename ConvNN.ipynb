# Convolutional Neural Networks

Let the input be $\large{X \in \mathbb{R}^{H \times W \times C_{in}}}$

Filter/Kernel be $\large{K \in \mathbb{R} ^ {k_h \times k_w \times C_{in}} }$ $\large{\text{ where, } K_{h}, K_{w} \text{ is smaller than input spatial dimension}}$

Stride $s$

Padding $p$

---

The output or the feature map has the following shape:

$$\Large{Y \in \mathbb{R}^{\lfloor \frac{H + 2p - k_H}{s} + 1 \rfloor \times \lfloor \frac{W + 2p - k_W}{s} + 1 \rfloor}}$$

---

For calculating the output of each pixel in the input space, the follwing formula applies. This is our standard formula for discrete convolution.

$$\Large{Y_{i,j} = \sum\limits_{u = 0}^{k_H - 1} \sum\limits_{v = 0}^{k_W - 1} \sum\limits_{c = 0}^{C_{in} - 1} X_{(i \cdot s + u),(j \cdot s + v), c} \cdot K_{u,v,c}}$$

Say, we use N filters in convolution layer, so the output is given as:

$$ \Large{Y \in \mathbb{R}^{H_{\text{out}} \times W_{\text{out}}\times N}} $$
