```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("MyFirstSparkApp").getOrCreate()
```

```python
spark.version
```

```python
data = [('Alice', 34), ('Bob', 45), ('Catherina', 29)]

columns = ["Name", "Age"]
```

```python
df = spark.createDataFrame(data, columns)
df
```
