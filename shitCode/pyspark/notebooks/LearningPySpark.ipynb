### Creating empty RDD with a schema

```python
from pyspark.sql.types import StructType, StructField, StringType
```

```python
mtSchemaRDD = StructType([StructField('firstName', StringType(), False), 
                         StructField('middleName', StringType(), True),
                         StructField('lastName', StringType(), False)]) ## True sets the nullable to true
```

```python
dfMTSchema = spark.createDataFrame(mtRDD, mtSchemaRDD) ## create new dataframe using using mtRDD and schema of mtSchemaRDD
```

```python
dfMTSchema.printSchema()
```

```python
dataframe2 = mtRDD.toDF(mtSchemaRDD) # Create df out of empty rdd using some schema
```

```python
dataframe2.show()
```

```python
df = spark.createDataFrame([(1,2.,'string1', date(2000,1,1), datetime(2000,1,1,12,0)),
                           (2,3., 'string2', date(2000,2,1),datetime(2000,2,1,12,0)),
                           (4,5., 'string3', date(2000,3,1), datetime(2000,3,1,12,0))], schema='a long, b double, c string, d date, e timestamp')
```

```python
df.schema
```

```python
df.printSchema()
```

```python
df.show(1, vertical=True)
```

```python
df.columns
```

```python
import pandas as pd
from pyspark.sql.functions import pandas_udf

@pandas_udf('long')
def pandas_plus_one(s:pd.Series) -> pd.Series:
    return s + 1

df.select(pandas_plus_one(df.a)).show()
```
