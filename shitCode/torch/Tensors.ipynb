```python
float32Tensor = torch.tensor([3.0, 4.0, 69.0], dtype=None, requires_grad=True)# Default dtype is float32
```

```python
float32Tensor
```

$\textbf{Important note: }$ 

`requires_grad` $\text{is a parameter that allows the autograd engine to later compute the gradients of the tensor with repect to a loss function menitoned in the program}$

This keeps track of all the operations in the tensor from that point onward

This enables us to just define the forward computation and PyTorch handles the gradients automatically. We don't have to write the backpropagation algoritm manually. We just call `y.backward()` where `y` is the tensor

When we use `torch.view()`, we just reshape the tensor. This effectively creates a new tensor. But the new tensor is referencing the same data as the original tensor(not a copy), so modifying the view tensor will also change the original tensor

```python
z = someObject.view(2,2,2)
```

```python
z
```

```python
torch.stack([z,z], dim=0)
```

```python
torch.squeeze(z)
```

There are two ways to index tensors

```python
print(z[1])
print(z[1][0])
print(z[0][0][1])
```

```python
print(z[:, :, :-1]) # Get all values of 1st and second dim, but exclude the last values in the 3rd dimension

print(z[:, :, -1:]) # Get all the values of 1st and 2nd dim, but only get the last value in the 3rd dimensio
```
