# RNNs From Scratch

$\text{Basically what you do is, make a network with hidden states. Those hidden states update with time step using the hidden state from the step before it and the currrent input.}$

$\text{So, suppose you have a sequence of inputs} x_1, x_2, x_3, \cdots , x_T . \text{Where T is the length of the input sequence}. \text{Let t be the sweeping parameter i.e time}$ 

### $\text{For each step, t, the following steps are to happen}$

1. #### Hidden State
   $h_t = \sigma{(W_h h_{t-1} + W_x x_t + b)}$

   $h_t$ $\text{is the hidden state at time t}$
   
   $h_{t-1}$ $\text{is the hidden state before time t}$

   $x_t$ $\text{is the current input i.e at time t}$

   $W_h$ $\text{is the weight matrix for the hidden state from the previous step}$

   $W_x$ $\text{is the weight matrix for the input at current time}$

   $b$ $\text{is the bias}$

   $\sigma$ $\text{is the activation function}$

### $\text{OUTPUT at time t}$

$y_t$ is the output at time t

$W_y$ is the weight matrix for generating the output from hidden state.

$b_y$ is the output bias
