# Support Vector Machines

Suppose we have $n$ labeled samples. 

$\Large{\{(x_i, y_i)\}_{i = 1}^n}$ , $\Large{ x_i \in \mathbf{R}^d, y_i \in \{-1, 1\} }$

#### Our goal is to find the optimal hyperplane that seperates two classes with the maximum margin.

- The classes being -1 and 1 in the set $y$. It can be dog or cat, cancer or no cancer and stuff.

- And the hyper plane is just the generalization of a flat decision boundary that seperates the feature space. In `2D` it is a line and in `3D` it is a plane.

> Mathematically, it is given as: &emsp;$ \Large{w^tx + b  = 0}$

> Where,

> w: The normal vector that defines the orientation

> x: Input vector
> b: Bias or offset from origin

- Margin is the distance between the hyperplane and the closest points from each class.
- Maximum margin means that we are looking for a hyper plane that is as far as possible from each class while also seperating them as accurately as possible

The constraints for this optimization problem is:

$\Large{y_i(w^Tx_i + b) >= 1}$ &ensp;**for all i**

# Duality Problem and Lagrange Multipliers

- *In a duality problem, the optimization problem at hand can be viewed from two perspectives, primal problem and/or the dual problem.*
- **`Weak duality:`** *The solution to primal is the upper bound to solution of dual and the solution to dual is the lower bound for primal*
- **`Strong Duality:`** *For convex optimization, the optimal values of primal and dual are the same. Or their duality gap is zero (under constraint qualification)*


For given function $f_0(x)$ , a task to minimize it for the constraints subject to $f_i(x) <= 0, i \in \{1,\ldots,m\}$ would be theoritically possible to program without the constraints if we do it by minimizing the following function:

$\Large{J(x) = f_0 (x) + \sum\limits_{i}I[f_i(x)]}$

Where, 

$I$ is the infinite step function: $ I[u] = 0$ if $u <= 0$, and $I[u] = Infinite$, otherwise

But $J(x)$ is hard to solve due to $I[u]$ being non-continuous. Hence, we approximate $I[u]$ by $\lambda u$ where, $\lambda$ is a positive constant. This yields the Lagrangian function:

$L(x, \lambda)$ $\Large{ = f_0(x) + \sum\limits_{i}\lambda _i f_i(x)}$

# Norm of a Vector

- Norm is a function that assigns a non-negative real number to a vector that represents the size or magnitude of the function
- The norm of a zero vector is zero and norm if zero for a vector if and only if it is a zero vector
- $L2$ Norm: Length of a vector in Euclidean space. Calculated as the square root of the sum of squares of the vector components
- $L1$ Norm: Sum of absolute values of the vector components

***For SVM, the objective function is to minimize the norm of the weights vector*** i.e

$\Large{\min\limits_{w,b} \frac{1}{2}\Arrowvert w\Arrowvert ^2}$

$\Large{s.t}$&emsp; $\Large{y_i(w^T x_i + b) >= 1, \forall i}$

We choose this constraint because

For $\Large{y_i = 1}$, $\Large{\vec{w} \cdot \vec{x} + b >= 1}$ 

And,

For, $\Large{y_i = -1}$, $\Large{\vec{w} \cdot \vec{x} + b <= 1}$

Now, the Lagrangian for this would be:

$\Large{\mathcal{L}(w, b, \alpha) = \frac{1}{2}\Arrowvert w \Arrowvert ^ 2 - \sum\limits_{i = 1}^{N} \alpha _i [y_i(w^T x_i + b) - 1]}$

Taking partial derivatives

$\Large{\frac{\partial{L}}{\partial{w}} = 0 =>  w = \sum\limits_{i = 1}^{N} \alpha _i y_i x_i} = 0$

$\Large{\frac{\partial{L}}{\partial{b}} = 0 => \sum\limits_{i = 1}^{N} \alpha _i y_i} = 0 $

We choose the above primal because 1/2 term and the square term helps mathematically, that we will discuss later.

The goal is to maximize $\frac{1}{\Arrowvert w \Arrowvert}$, which is equal to minimizing $\frac{1}{2}\Arrowvert w \Arrowvert ^ 2$

Now, for some cases that might not be linearly seperable. We introduce the slack variable. This is done so that the hyperplane does not strictly try to seperate the classes. And also, that would not be possible with non linear cases. That slack variable is $\varepsilon$ or for $i$ data points, $\varepsilon_i$

Now, think intuitvely, we want to slacken the hard margin. So, for that we need to slacken the constraint, i.e.

$\Large{y_i(\vec{w} \cdot \vec{x} + b) >= 1 - \varepsilon_i}$ &emsp;, $ \Large{\varepsilon_i >= 0} $

With that, the objective function becomes:

$\Large{\min\limits_{\vec{w}, x, \varepsilon} \frac{1}{2}\Arrowvert w \Arrowvert ^ 2 + C \sum\limits_{i = 1}^{n}\varepsilon_i}$

$\Large{C}$ is a regularization parameter that controls how much is the misclassification is penalized. So for bigger $\Large{C}$, the SVM tries harder to reduce slack and for smaller $\Large{C}$, the SVM will tolorate some misclassifications

Now, we make the Lagrangian again, but with the soft margin consideration

$\Large{\mathcal{L}(\vec{w}, b, \varepsilon, \alpha) = \frac{1}{2}\Arrowvert w \Arrowvert ^ 2 + C \sum\limits_{i = 1}^{n}\varepsilon_i - \sum\limits_{i = 1}^{n}\alpha_i (y_i(\vec{w} \cdot \vec{x} + b) - 1 + \varepsilon_i )}$

Now, we form the dual by partially differntiating the Lagrangian function with respect to $\vec{w}, b, \varepsilon_i$ and setting them to zero

1. $\Large{\frac{\partial{\mathcal{L}}}{\partial{w}} = w + 0 - \sum\limits_{i = 1}^{n} (\alpha_i y_i \vec{x}) = 0 => w = \sum\limits_{i = 1}^{n} (\alpha_i y_i \vec{x})}$

2. $\Large{\frac{\partial{\mathcal{L}}}{\partial{b}} = 0 + 0 - \sum\limits_{i = 1}^{n} \alpha_i y_i = 0 => \sum\limits_{i = 1}^{n}\alpha_i y_i =0}$

3. $\Large{\frac{\partial{\mathcal{L}}}{\partial{\varepsilon_i}} = 0 + C * 1 - \alpha_i = 0 => C - \alpha_i =0}$&emsp;&emsp;

---
Remember, that in the third differentiation, we are differentiating with respect to the term $\varepsilon_i$ and not a single $\varepsilon$. Hence, the summation term vanishes

- **To handle a case of non-linear classification, we just move the data to higher dimension using kernels. Most common kernel I have seen is the polynomial kernel.**
- **We are sure to find a higher dimension where the data is linearly seperable, due to Cover's theorem which states that:**

>*`A complex pattern recognition problem cast in a high dimension space non-linearly is more likely to be linearly seperable than in a low-dimensional space`*

- **The kernel trick is that we don\'t transform the data from lower dimension to higher dimension, but we just compute the inner products in that higher dimension**

```python
from sklearn import datasets
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, accuracy_score
```

```python
iris = datasets.load_iris()

X = iris.data
y = iris.target

y_binary = (y != 0).astype(int)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_binary, test_size=0.2, random_state=69)
```

```python
svm_model = SVC(kernel='linear', C = 1.0)
svm_model.fit(X_train, y_train)
```

```python
y_predict = svm_model.predict(X_test)

print(f"Accuracy: {accuracy_score(y_test, y_predict)}")
print(classification_report(y_test, y_predict, target_names = ['setosa', 'Non-Setosa']))
```
